```
import numpy as np
from rfsoc_awg import *
from rfsoc_dashboard import AWGDashboard, DemoScope

DRY_RUN = False        # True = no board needed: simulated DAC/ADC to play with the dashboard
```

```
# Same start-up sequence as your adc_scope notebook.
if DRY_RUN:
    awg = AWG(window=1 * us, dry_run=True, verbose=False)
    scope = DemoScope(awg)                       # simulated ADC
else:
    from pynq import Overlay, PL
    import xrfclk
    from adc_scope import AdcScope

    PL.reset()
    xrfclk.set_ref_clks(lmk_freq=245.76, lmx_freq=491.52)
    ol = Overlay("awg_adc.bit")

    awg = AWG(window=1 * us, bitfile=ol, verbose=False)   # window = length of one AWG slot
    _ = awg.connect()
    scope = AdcScope(ol)
```

```
dash = AWGDashboard(awg, scope=scope)     # adc_vfs=<volts> for volts/dBm, template="plotly_dark" for dark mode
dash.ui
```
